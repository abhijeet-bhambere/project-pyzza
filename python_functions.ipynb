## Multiple parameters

```python
#  The function accepts two arguments
def raise_to_power(value1, value2):
    """Raise value1 to the power of value2."""
    return value1 ** value2
```

## Nested Functions
Its essentially one or more inner function(s) defined within an outer , enclosing function. **python follows the LEGB principle**

In an inner function, if a variable is not found within itself, the code will look for the variable in the outer function. 

As per design , python will look for the variable first locally , then globally & lastly within built-ins

```python
def mod2plus5(x1, x2, x3):
    """Returns the remainder plus 5 of three values."""
    def inner(x):
        """Returns the remainder plus 5 of a value."""
        return x % 2 + 5
    # Calling the inner function to serve as the output of the outer function
    return inner(x1), inner(x2), inner(x3)
```

```python
mod2plus5(1, 2, 3)
```

## A concept called **closure**.
The outer function returns the output of the inner function as an output. 

In below example: primary purpose of `raise_val` is not to perform a calculation directly, but **to create and return another function** (the inner function). The `n` argument passed to `raise_val` is "captured" by this inner function.

**The closure part:** Even though `inner` is defined inside raise_val, and `n` is a parameter of `raise_val`, **the inner function "remembers" the value of `n`** from the environment in which it was created. **This "remembering" of the n value, even after raise_val has finished executing, is what a closure is.**

```python
def raise_val(n):
    """Return the inner function."""

    def inner(x):
        """Raise x to the power of n."""
        raised = x ** n
        return raised

    return inner
```

```python
square = raise_val(2)
cube = raise_val(3)
```

```python
square(4)
```

```python
cube(6)
```

## Use of `nonlocal`
Used to delimit the scope of a variable. Changes the value in the enclosing scope as well.

The `nonlocal` keyword modifies the variables defined in the enclosing scope of a nested function.

It kind of upgrades the scope of a variable.

```python
def outer():
    """Prints the value of n."""
    n = 1
    def inner():
        nonlocal n
        n = 2
        print(n)
    inner()
    print(n)
```

```python
outer()
```

Another example:

```python
# Define echo_shout()
def echo_shout(word):
    """Change the value of a nonlocal variable"""
    
    # Concatenate word with itself: echo_word
    echo_word = word + word
    
    # Print echo_word
    print(echo_word)
    
    # Define inner function shout()
    def shout():
        """Alter a variable in the enclosing scope"""    
        # Use echo_word in nonlocal scope
        nonlocal echo_word
        
        # Change echo_word to echo_word concatenated with '!!!'
        echo_word = echo_word + "!!!"
    
    # Call function shout()
    shout()
    
    # Print echo_word
    print(echo_word)

# Call function echo_shout() with argument 'hello'
echo_shout("hello")
```

## Default arguments

```python
# Add a default argument to a function
def power(num, pow=1):
    """Raise num to the power defined by 'pow' (default pwer is 1)"""
    new_val = num**pow
    return new_val
```

```python
# The func returns output with the value raised to default power of 1
power(2)
```

```python
power(2,4)
```

## Flexibile arguments
The function will accept any no. of arguments that do not have tobe defined in the function definition.

```python
def add_all(*args):
    """Sum all values in *args together."""
    # Initialize sum
    sum_all = 0
    # Accumulate the sum
    for num in args:
        sum_all += num
    return sum_all
```

```python
add_all(1)
```

```python
add_all(1,9,10,7)
```

Another example
### Keyword arguments allow for passing `key:value` pairs

```python
def num_kwargs(**kwargs):
    """Returns the number of keyword arguments passed in **kwargs"""
    for key, value in kwargs.items():
     print(key + ": " + value)
```

```python
num_kwargs(a="1" , b="2")
```

```python
# Define shout_echo
def shout_echo(word1, echo=1):
    """Concatenate echo copies of word1 and three
     exclamation marks at the end of the string."""

    # Concatenate echo copies of word1 using *: echo_word
    echo_word = word1 * echo

    # Concatenate '!!!' to echo_word: shout_word
    shout_word = echo_word + '!!!'

    # Return shout_word
    return shout_word

# Call shout_echo() with "Hey": no_echo
no_echo = shout_echo("Hey")

# Call shout_echo() with "Hey" and echo=5: with_echo
with_echo = shout_echo("Hey",5)

# Print no_echo and with_echo
print(no_echo)
print(with_echo)
```

# Lambda functions
Lambda functions in Python are small, anonymous functions defined using the `lambda` keyword. They are typically used for short, simple operations where a full function definition using `def` is unnecessary.


Syntax: `lambda arguments: expression`

**Use Cases:**
- **Short, simple operations:** When you need a function for a single, straightforward task.
- **Higher-order functions:** Passing functions as arguments to map, filter, sorted, etc.
- **Inline functions:** Defining functions directly where they are needed.


```python
raise_to_power = lambda x, y: x ** y
```

```python
raise_to_power(2,3)
```

Another example for echo:

```python
echo_word = lambda word1,echo: word1*echo
```

```python
echo_word("hello",4)
```

## Anonymous functions:
- Using `map` to apply a lambda function to a sequence (example a list)
    - Notice that the function is created AND appllied inline. Its not stored in the environment -- meaning it cannot be called anywhere else in the space

- The `filter()` function in Python is a built-in function that is used to filter elements from an iterable (like a list, tuple, or set) based on a given condition. 
    - It takes two arguments: a function and an iterable. The function is used to test each element in the iterable, and if the function returns True for an element, that element is included in the resulting iterator

- The `reduce()` function is useful for performing some computation on a list & **returns a single value as a result**.
    - To use reduce(), you must import it from the functools module.

Using the `filter` function
 

```python
alist = [2,3,4,5,6]
```

```python
ld_list = map(lambda x: x**2, alist)
```

```python
print(list(ld_list))
```

```python
2 in alist
```

```python
filter(lambda a: a > 3, alist)
```

```python
print(list(filter(lambda a: a > 3, alist)))
```

```python
aa = filter(lambda x: len(x) > 5, ['hello', 'world', 'python', 'is', 'awesome'])
```

Using `reduce` function

```python
# Create a list of strings: stark
stark = ['robb', 'sansa', 'arya', 'brandon', 'rickon']

from functools import reduce

# Use reduce() to apply a lambda function over stark: result
result = reduce(lambda item1, item2: item1+item2, stark)

# Print the result
print(result)
```

## Error-handling
Defining what a function does when incorrect arguments are passed while calling it. The error message should help user to identify what is missing / incorrect & make corrections accordingly.

'_Catching_' an exception using `try-except` : 
- Attempt to run the code in the `try` block, if all Ok then jump over the `except` block to the next line in the code
- If there's an exception (i.e the `try` block fails to execute) then execute the `except` block

### Some common error types are:
| Error type | Description |
|------------|-------------|
| `TypeError` | when argument has invalid object type|
| `ValueError` | When argument has valid object type but inappropriate value|
| `ZeroDivisionError` | Passing 0 to a division operation |
| `IndexError` | Accessing an index that is out of the bounds of a list or tuple |
| `NameError` | Refers to using a variable that has not been defined|

```python
# Using the error -handling
def sqrt_excp(x):
    """Returns the square root of a number"""
    try:
        return x**0.5
    except:
        print("x must be an int or float")
```

```python
sqrt_excp("hello")
```

```python
# Using the error -handling -- pt2
def sqrt_excp1(x):
    """Returns the square root of a number"""
    if x < 0:
        raise ValueError("x must be non-negative")
    try:
        return x**0.5
    except TypeError:
        print("x must be an int or float")
```

```python
sqrt_excp1(-12)
```

```python
len(2)
```
